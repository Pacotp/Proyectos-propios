En este cuaderno vamos a hacaer una simulación de un proyecto de los datos de lanzamiento de un cohete

```python
import numpy as np
import matplotlib.pyplot as plt
```

Los datos que vamos a utilizar son la siguiente matriz de datos donde cada columna representa un sensor y cada fila es la temperatura en el tiempo i=1,2,3

```python
temp_motores = np.array([
    [320.5, 315.2, 330.0, 310.8],
    [340.1, 345.8, 350.2, 320.1],
    [410.2, 395.0, 420.4, 315.4],  # Nota el incremento aquí
    [480.9, 490.1, 510.5, 310.2],  # Temperaturas críticas
    [250.0, 245.5, 260.1, 200.8]   # Fase de enfriamiento
], dtype=np.float32)
```

Los datos que tenemos que tener en cuenta es la temperatura critica de peligro es de 400.0 grados

```python
promedios = np.mean(temp_motores, axis=0)
print("Temperaturas promedio por motor:", promedios)
```

```python
#Ahora vamos a calcular la teperatura maxima global de todos los motores
maxima_global = np.max(temp_motores)
print("Temperatura máxima global alcanzada por todos los motores es:", maxima_global)
```

Ahora vamos a ver que elementos cumplen la restriccion de entrar en situacion de peligro cuando igualan o superan los 400 grados

```python
mascara =temp_motores>=400
alarma = temp_motores[mascara]
print("Los motores que han alcanzado la temperatura critica son: ", alarma)
```
